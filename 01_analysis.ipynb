# Support Vector Regression for Financial Time Series
## A Quantitative Research Framework

**Asset**: Apple Inc. (AAPL) 2015-2024 (~2,500 trading days)  
**Framework**: Walk-forward validation | No lookahead bias | Transaction costs included

---

### Research Objectives

1. Derive and implement the epsilon-SVR optimisation from first principles
2. Engineer 40+ financial features (technical, microstructure, regime)
3. Compare RBF / Linear / Polynomial kernels and tune via TimeSeriesSplit
4. Run a walk-forward backtest with transaction costs and volatility-targeting position sizing
5. Benchmark SVR against a stacking ensemble (SVR + XGBoost + LightGBM + RandomForest)
6. Apply formal statistical tests: Diebold-Mariano, Pesaran-Timmermann, bootstrap Sharpe CIs
7. Analyse performance across low-vol / high-vol market regimes

> **Quant Insight**: In efficient markets R^2 near 0 for next-day returns is *expected*, not a failure. The relevant metric is *risk-adjusted excess return* after costs.
